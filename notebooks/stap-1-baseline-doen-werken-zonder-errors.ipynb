```python
#baseline uitvoeren poging 2: 

from pathlib import Path
import subprocess
import sys

# Zoek het juiste train.csv-bestand
matches = list(Path("/kaggle/input").rglob("train.csv"))

if not matches:
    raise FileNotFoundError("train.csv is niet gevonden!")

DATA_DIR = matches[0].parent

print("Gebruikte datasetmap:", DATA_DIR)
print("Aanwezige bestanden:", [p.name for p in DATA_DIR.iterdir()])

# Zoek het baseline-script
scripts = list(Path("/kaggle/input").rglob("hms_sklearn_baseline.py"))

if not scripts:
    raise FileNotFoundError("Baseline-script niet gevonden!")

SCRIPT = scripts[0]

# Voer de baseline uit
subprocess.run([
    sys.executable,
    str(SCRIPT),
    "--data-dir", str(DATA_DIR),
    "--mode", "quick",
    "--folds", "3",
    "--n-jobs", "2"
], check=True)
```
